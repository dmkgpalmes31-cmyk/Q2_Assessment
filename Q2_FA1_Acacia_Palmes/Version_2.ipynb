```python
# Ask the user to enter two numbers
num1 = float(input("Enter the number 1: "))

num2 = float(input("Enter the number 2: "))

# Display the available operations
print("1 - Addition")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Division")

# Ask the user to choose an operation
mode = (int(input("Choose operation: ")))

# Function for addition
def add_numbers(num1, num2):
    return num1 + num2

# FuncTion for subtraction
def subtract_numbers(num1, num2):
    return num1 - num2

# Function for multiplication
def multiply_numbers(num1, num2):
    return num1 * num2

# Function for division
def divide_numbers(num1, num2):
    if num2 == 0:
       return "Cannot divide by zero"
    return num1 / num2

# Perform the selected operation and display the result
if mode == 1:
   print(add_numbers(num1, num2))
elif mode == 2:
   print(subtract_numbers(num1, num2))
elif mode == 3:
   print(multiply_numbers(num1, num2))
elif mode == 4:
   print(divide_numbers(num1, num2))
elif mode < 1 or mode > 4:
   print("Invalid Choice")
```

Short Reflection

1. I created four functions, which are add_numbers(), subtract_numbers(), multiply_numbers(), and divide_numbers().
2. My functions used num1 and num2 as my parameters
3. The arguments passed were the two numbers entered by the user, num1 and num2.
4. My program uses print() to display the value returned by the selected function.
5. Using functions makes the code more organized, readable, and easier to understand.
